```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer

# Cargar los datos
data = load_breast_cancer()
# Crear un DataFrame
df = pd.DataFrame(data.data, columns=data.feature_names)
# Agregar la variable objetivo
df["target"] = 1 - data.target
```

```python
correlacion = df.drop(columns="target").corrwith(df["target"],method="pearson")
print(correlacion)
```

```python
corr_abs=abs(correlacion)
mejor = corr_abs.idxmax()          # nombre de la característica
valor_abs = corr_abs.max()         # correlación absoluta máxima
valor_real = correlacion[mejor]    # con su signo original

print(f"Mayor correlación: {mejor} ({valor_real:.3f})")
```

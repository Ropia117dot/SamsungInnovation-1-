### Ejercicio 5 · Escalado a mano

Escala `total_bill` con **min-max** y con **z-score**, sin usar scikit-learn. Comprueba con `describe()` la media, la desviación, el mínimo y el máximo de cada versión.

````python
tb=propinas["total_bill"]

escalado=pd.DataFrame({
    "original":tb,
    "min_max":(tb - tb.min())/(tb.max() - tb.min()),
    "z_score":(tb - tb.mean())/tb.std()
})

escalado.describe().loc[["mean","std","min","max"]]
````

### Ejercicio 6 · Logaritmo

a) Calcula la asimetría de `total_bill` y de `np.log1p(total_bill)`.

b) Dibuja sus dos histogramas, uno al lado del otro.

````python
asimetria=np.log1p(propinas["total_bill"])

fig,axes= plt.subplots(1,2)
sns.histplot(propinas["total_bill"],bins=50,ax=axes[0])
axes[0].set_title(f"Original asimetria = {propinas["total_bill"].skew():.2f}")
sns.histplot(asimetria,bins=50,ax=axes[1])
axes[1].set_title(f"Con log1p = {asimetria.skew():.2f}")
plt.tight_layout()
plt.show()
````
